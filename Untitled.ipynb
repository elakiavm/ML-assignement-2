```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
data = pd.read_csv('TrainDataset2024.csv')

# Display the first few rows to understand the structure
print(data.head())
```

```python
data.isnull().sum()
```

```python
# Check for missing values and data types
print(data.info())

# Summary statistics for numerical columns
print(data.describe())
```

```python
data.isnull().sum()
```

```python
# Visualize missing values with a heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(data.isnull(), cbar=False, cmap="viridis")
plt.title("Missing Data Heatmap")
plt.show()

# Impute missing values (example: fill numerical columns with the median and categorical with the mode)
for column in data.columns:
    if data[column].dtype == "object":
        data[column].fillna(data[column].mode()[0], inplace=True)
    else:
        data[column].fillna(data[column].median(), inplace=True)

# Confirm there are no missing values left
print(data.isnull().sum())
```

```python
# Box plot for all numerical columns
plt.figure(figsize=(12, 8))
sns.boxplot(data=data.select_dtypes(include=[np.number]))
plt.xticks(rotation=90)
plt.title("Box Plot for Numerical Features")
plt.show()
```

```python
def analyze_999_values(df):
    # Count 999 values in each column
    missing_counts = (df == 999).sum()
    
    # Calculate percentage of 999 values
    missing_percentage = (df == 999).sum() * 100 / len(df)
    
    # Create summary DataFrame
    missing_summary = pd.DataFrame({
        'Count of 999s': missing_counts,
        'Percentage of 999s': missing_percentage.round(2)
    })
    
    # Filter only columns that have 999s
    missing_summary = missing_summary[missing_summary['Count of 999s'] > 0]
    
    return missing_summary.sort_values('Count of 999s', ascending=False)
```

```python
# Display missing value analysis
missing_analysis = analyze_999_values(data)
print("\n=== Analysis of 999 Values (Missing Data) ===")
print(missing_analysis)

# Step 6: Visualize the distribution of 999 values
plt.figure(figsize=(12, 6))
missing_analysis['Count of 999s'].plot(kind='bar')
plt.title('Number of 999 Values (Missing Data) by Feature')
plt.xlabel('Features')
plt.ylabel('Count of 999s')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Step 7: Summary statistics
total_999s = (data == 999).sum().sum()
total_cells = data.size
percentage_999s = (total_999s / total_cells) * 100

print("\n=== Overall Missing Data Summary ===")
print(f"Total number of 999 values: {total_999s}")
print(f"Total number of cells in dataset: {total_cells}")
print(f"Percentage of missing data: {percentage_999s:.2f}%")

# Step 8: Display features with high missing rates (>5%)
high_missing = missing_analysis[missing_analysis['Percentage of 999s'] > 5]
print("\n=== Features with High Missing Rates (>5%) ===")
print(high_missing)
```
